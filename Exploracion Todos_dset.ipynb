```python
#Otro intento

# ============================================================
# DESCOMPOSICIÓN TRANSVERSAL DE I = I_0 + I_{-1,1}
# Una sola barra por cada N0
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import os
import zipfile
from google.colab import files

# ============================================================
# CONFIGURACIÓN GENERAL DE ESTILO PARA ARTÍCULO
# ============================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "legend.fontsize": 10,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "mathtext.fontset": "cm",
    "text.usetex": False
})

sns.set_style("whitegrid")

archivos_generados = []

# ============================================================
# PARÁMETROS
# ============================================================

TOTAL_AGENTES = 1500
WIDTH = 40
HEIGHT = 40
V = WIDTH * HEIGHT - TOTAL_AGENTES

# Umbral teórico N_1 <= N_0 + V
N0_UMBRAL = (TOTAL_AGENTES - 2 * V) / 3

DIBUJAR_UMBRAL = True

# Colores más suaves
COLOR_I0 = "#D9B44A"        # dorado suave
COLOR_IEXT = "#C75C5C"      # rojo suave
COLOR_TOTAL = "#222222"     # casi negro, pero solo para la línea

# ============================================================
# FUNCIÓN PARA TOMAR SOLO EL ESTADO FINAL POR N0
# ============================================================

def preparar_datos_transversales(df_item):
    """
    Toma un DataFrame que puede contener varios pasos temporales.
    Devuelve una sola fila por cada N0=N2, usando el último step disponible.
    """

    df = df_item.copy()

    # Ordenamos por N2 y step si existe
    if "step" in df.columns:
        df = df.sort_values(["N2", "step"])

        # Si hay varias corridas para el mismo N2, primero tomamos el último
        # paso de cada corrida y luego promediamos por N2.
        if "corrida" in df.columns:
            df_final = (
                df.groupby(["N2", "corrida"], as_index=False)
                  .tail(1)
                  .reset_index(drop=True)
            )

            numeric_cols = df_final.select_dtypes(include=[np.number]).columns
            df_final = (
                df_final.groupby("N2", as_index=False)[numeric_cols]
                        .mean()
            )
        else:
            df_final = (
                df.groupby("N2", as_index=False)
                  .tail(1)
                  .reset_index(drop=True)
            )
    else:
        # Si no hay columna step, asumimos que ya es un resumen transversal.
        df_final = df.sort_values("N2").reset_index(drop=True)

    return df_final


# ============================================================
# GRÁFICAS POR CADA VALOR DE TAU
# ============================================================

for i, df_item in enumerate(df_list):

    if df_item.empty:
        continue

    df_plot = preparar_datos_transversales(df_item)

    df_plot = df_plot.sort_values("N2").reset_index(drop=True)

    tol_value = df_plot["tol"].iloc[0]
    tol_str = str(tol_value).replace(".", "_")

    # ========================================================
    # Componentes de I
    # ========================================================
    # A = 1, B = 0, C = -1
    #
    # I_0 = C_{0,1} + C_{0,-1}
    #     = C_AB + C_BC
    #
    # I_{-1,1} = 2 C_{1,-1}
    #          = 2 C_AC

    df_plot["I_0"] = df_plot["C_AB"] + df_plot["C_BC"]
    df_plot["I_extremos"] = 2 * df_plot["C_AC"]
    df_plot["I_total"] = df_plot["I_0"] + df_plot["I_extremos"]

    print(f"\nTau = {tol_value}")
    print(f"Filas graficadas: {len(df_plot)}")
    print(df_plot[["N2", "I_0", "I_extremos", "I_total"]].head())

    # ========================================================
    # Figura
    # ========================================================

    fig, ax = plt.subplots(figsize=(8, 5.2))

    x = df_plot["N2"]

    # Ancho automático de barras
    if len(x) > 1:
        paso_N0 = np.min(np.diff(sorted(x.unique())))
        ancho_barra = 0.65 * paso_N0
    else:
        ancho_barra = 60

    ax.bar(
        x,
        df_plot["I_0"],
        width=ancho_barra,
        label=r"$I_0=C_{0,1}+C_{0,-1}$",
        color=COLOR_I0,
        alpha=0.80,
        edgecolor="white",
        linewidth=0.5
    )

    ax.bar(
        x,
        df_plot["I_extremos"],
        width=ancho_barra,
        bottom=df_plot["I_0"],
        label=r"$I_{-1,1}=2C_{1,-1}$",
        color=COLOR_IEXT,
        alpha=0.80,
        edgecolor="white",
        linewidth=0.5
    )

    # Línea del total
    ax.plot(
        x,
        df_plot["I_total"],
        "o-",
        color=COLOR_TOTAL,
        linewidth=1.4,
        markersize=4,
        label=r"$I=I_0+I_{-1,1}$"
    )

    # Umbral teórico opcional
    if DIBUJAR_UMBRAL:
        ax.axvline(
            N0_UMBRAL,
            color="gray",
            linestyle="--",
            linewidth=1.2,
            alpha=0.65
        )

        ax.text(
            N0_UMBRAL,
            ax.get_ylim()[1] * 0.95,
            r"$N_1=N_0+V$",
            rotation=90,
            va="top",
            ha="right",
            fontsize=10,
            color="gray"
        )

    ax.set_title(rf"Cross-sectional decomposition of $I$, $\tau={tol_value}$")
    ax.set_xlabel(r"$N_0$")
    ax.set_ylabel(r"Dissatisfaction")

    ax.legend(
        fontsize=10,
        frameon=True,
        loc="upper right"
    )

    ax.grid(axis="y", alpha=0.25)
    ax.grid(axis="x", visible=False)

    plt.tight_layout()

    # ========================================================
    # Guardar figuras
    # ========================================================

    nombre_png = f"figura_descomposicion_I_tau_{tol_str}_clara.png"
    nombre_pdf = f"figura_descomposicion_I_tau_{tol_str}_clara.pdf"

    plt.savefig(nombre_png, dpi=300, bbox_inches="tight")
    plt.savefig(nombre_pdf, bbox_inches="tight")

    plt.close(fig)

    archivos_generados.append(nombre_png)
    archivos_generados.append(nombre_pdf)

# ============================================================
# Crear ZIP
# ============================================================

zip_name = "figuras_descomposicion_I_tau_claras.zip"

with zipfile.ZipFile(zip_name, "w") as zipf:
    for archivo in archivos_generados:
        zipf.write(archivo)

print("\nArchivos generados:")
for archivo in archivos_generados:
    print(archivo)

files.download(zip_name)
```
