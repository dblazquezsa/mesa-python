```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import random

def ejecutar_modelo_schelling(tamano_grid=500, proporcion_vacia=0.1, umbral_similitud=0.4, max_iteraciones=100):
    """
    Ejecuta el modelo de segregación de Schelling para 3 poblaciones.

    0: Espacio vacío
    1: Población A (Rojo)
    2: Población B (Azul)
    3: Población C (Verde)
    """
    # Definir la probabilidad de cada estado inicial
    p_vacia = proporcion_vacia
    p_pob = (1 - proporcion_vacia) / 3
    probabilidades = [p_vacia, p_pob, p_pob, p_pob]

    # Inicializar el grid
    grid = np.random.choice([0, 1, 2, 3], size=(tamano_grid, tamano_grid), p=probabilidades)
    grid_inicial = np.copy(grid)

    def obtener_celdas_vacias(matriz):
        return list(zip(*np.where(matriz == 0)))

    def es_feliz(matriz, x, y, umbral):
        agente = matriz[x, y]
        if agente == 0:
            return True # Los espacios vacíos no tienen estado de felicidad

        iguales = 0
        total_vecinos = 0

        # Evaluar la vecindad de Moore (8 vecinos que rodean la celda)
        for i in [-1, 0, 1]:
            for j in [-1, 0, 1]:
                if i == 0 and j == 0:
                    continue
                nx, ny = x + i, y + j

                # Verificar límites del grid
                if 0 <= nx < tamano_grid and 0 <= ny < tamano_grid:
                    vecino = matriz[nx, ny]
                    if vecino != 0:
                        total_vecinos += 1
                        if vecino == agente:
                            iguales += 1

        # Si un agente está completamente aislado, se asume feliz
        if total_vecinos == 0:
            return True

        # Condición de satisfacción
        return (iguales / total_vecinos) >= umbral

    # Bucle principal de iteraciones
    for iteracion in range(max_iteraciones):
        agentes_infelices = []

        # Identificar a todos los agentes que no cumplen su umbral de similitud
        for i in range(tamano_grid):
            for j in range(tamano_grid):
                if grid[i, j] != 0 and not es_feliz(grid, i, j, umbral_similitud):
                    agentes_infelices.append((i, j))

        # Si no hay agentes infelices, el sistema alcanzó el equilibrio
        if not agentes_infelices:
            print(f"Equilibrio topológico alcanzado en la iteración {iteracion}.")
            break

        celdas_vacias = obtener_celdas_vacias(grid)

        # Mezclar aleatoriamente para evitar sesgos direccionales en la reubicación
        np.random.shuffle(agentes_infelices)
        np.random.shuffle(celdas_vacias)

        # Mover la cantidad máxima posible de agentes infelices a espacios vacíos
        movimientos = min(len(agentes_infelices), len(celdas_vacias))
        for i in range(movimientos):
            pos_vieja = agentes_infelices[i]
            pos_nueva = celdas_vacias[i]

            # Reubicación
            grid[pos_nueva] = grid[pos_vieja]
            grid[pos_vieja] = 0

    return grid_inicial, grid

# --- Parámetros ajustables ---
TAMANO = 150
VACIOS = 0.10
UMBRAL = 0.7 # El agente quiere que al menos el 40% de sus vecinos sean de su mismo grupo
ITERACIONES = 150

# Ejecutar la simulación
inicial, final = ejecutar_modelo_schelling(
    tamano_grid=TAMANO,
    proporcion_vacia=VACIOS,
    umbral_similitud=UMBRAL,
    max_iteraciones=ITERACIONES
)

# --- Visualización ---
cmap = ListedColormap(['white', '#e63946', '#457b9d', '#2a9d8f']) # Vacío, A, B, C

fig, ax = plt.subplots(1, 2, figsize=(14, 7))

# Gráfico Estado Inicial
ax[0].imshow(inicial, cmap=cmap, interpolation='nearest')
ax[0].set_title('Estado Inicial (Aleatorio)')
ax[0].axis('off')

# Gráfico Estado Final
ax[1].imshow(final, cmap=cmap, interpolation='nearest')
ax[1].set_title(f'Estado Final (Umbral: {UMBRAL*100}%)')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```
